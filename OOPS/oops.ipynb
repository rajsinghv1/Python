**Constructor --init--Function**

```python
class Student:
    def __init__(self, name, marks):
        self.name = name
        self.marks = marks
        print("adding new student in Database..")

S1 = Student("Karan", 97)
print(S1.name, S1.marks)

S2 = Student("arjun", 98)
print(S2.name, S2.marks)
```

```python
class Student:

    #default constructions
    def __init__(self):
        pass

        #parameterized constructions
    def __init__(self, name, marks):
         self.name = name
         self.marks = marks
         print("adding new student in Database..")

S1 = Student("Karan", 97)
print(S1.name, S1.marks)

S2 = Student("arjun", 98)
print(S2.name, S2.marks)
```
