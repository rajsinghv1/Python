### Map, Filter and Reduce

**Map**

```python
def cube(x):
    return x * x * x

print(cube(2))
l = [1, 2, 4, 6, 4, 3]
newl = (map(cube, l))
print(list(newl))
```

```python
def addition(n):
    return n + n

# We double all numbers using map()
numbers = (1, 2, 3, 4)
result = map(addition, numbers)
print(list(result))
```

```python
# List of numbers
numbers = [1, 2, 3, 4, 5]

# Double each number using the map function
doubled = map(lambda x: x * 2, numbers)

# Print the doubled numbers
print(list(doubled))
 
```
