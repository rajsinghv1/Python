```python
def double(x):
  return x * 2
print(double(5))
```

```python
double = lambda x: x*2
cube = lambda x: x*x*x
print(double(5))
print(cube(5))
```

```python
double = lambda x: x*2
cube = lambda x: x*x*x
avg = lambda x, y: (x + y) / 2
print(double(5))
print(cube(5))
print(avg(3, 5))
```

```python
double = lambda x: x*2
cube = lambda x: x*x*x
avg = lambda x, y, z: (x + y + z) / 3
print(double(5))
print(cube(5))
print(avg(3, 5, 10))
```
