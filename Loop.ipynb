### Loop in python

```python
name = 'Abhishek' # string
for i in name:
    print(i)
    if(i =="b"):
        print("This is something special!")
```

```python
colors = ["Red", "Green", "Blue", "Yellow"]
for color in colors:
    print(color)
    for i in color:
        print(i)
```

### While loop

```python
i = 0
while(i<3):
    print(i)
    i = i + 1
```

### Do while loop

```python
i = 0
while True:
    print(i)
    i = i + 1
    if(i%4 == 0):
        break
```

### range():

```python
for k in range(5):
    print(k)
```

### Break Statements

```python
for i in range(12):
    if(i == 10):
        break
    print("5 X", i+1, "=", 5 * (i+1))
        
print("Loop ko chodkar nikal gaya")
```

### Continues Statements

```python
for i in range(12):
    if(i == 10):
        print("Skip the iteration")
        continue
print("5 X", i, "=", 5 * i)
```
