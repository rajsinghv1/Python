### Python Variables
In Python, variables are created when you assign a value to it:

```python
x = 5
y = "Hellow, world!"
print(x)
print(y)
```

```python
x = 5
y = "John"
print(x)
print(y)
```

```python
x = 4       # x is of type int
x = "Sally" # x is now of type str
print(x)
```

```python
# Casting
# If you want to specify the data type of a variable, this can be done with casting.

# Example
x = str(3)    # x will be '3'
y = int(3)    # y will be 3
z = float(3)  # z will be 3.0
print(x)
print(y)
print(z)
```

```python
x = 5
y = "John"
print(type(x))
print(type(y))
```

```python
x = 4
y = "RajSingh"
print(type(x))
print(type(y))
```

### Python Variables - Assign Multiple Values

```python
x, y, z = "Orange", "Banana", "Cherry"
print(x)
print(y)
print(z)
```

```python
x = y = z = "Orange"
print(x)
print(y)
print(z)
```

```python
fruits = ["apple", "banana", "cherry"]
x, y, z = fruits
print(x)
print(y)
print(z)
print(fruits)
```
