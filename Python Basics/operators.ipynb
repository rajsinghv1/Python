### Operators

#### Arithmatic Operator

```python
#arithmatic opertors
a=5
b=2

print(a+b)
print(a-b)
print(a*b)
print(a/b)
print(a%b) #remainder
print(a ** b) #a^b
```

```python
#relational operators means true and false
a = 50
b = 20
print(a == b)
print(a != b)
print(a >= b)
print(a <= b)
```

```python
#Assignment opertors
num = 10
num += 10
print("num :", num)
```
