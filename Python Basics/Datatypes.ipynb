### Data Type

- Text Type:	str
- Numeric Types:	int, float, complex
- Sequence Types:	list, tuple, range
- Mapping Type:	dict
- Set Types:	set, frozenset
- Boolean Type:	bool
- Binary Types:	bytes, bytearray, memoryview
- None Type:	NoneType

```python
## Integer Example
age=35
type(age)
```

```python
##floating point datatype
height=5.11
print(height)
print(type(height))
```

```python
## string datatype example
name="Raj"
print(name)
print(type(name))
```

```python
## boolean datatype
is_true=True
type(is_true)
```

```python
a=10
b=10

type(a==b)
```
